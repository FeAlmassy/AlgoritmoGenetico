# Diabetes stages (3 classes)

The model learns to tell whether a person has **no diabetes (class 0)**, **prediabetes (class 1)** or **diabetes (class 2)**, using health and lifestyle data from a survey.

The file uses a random sample of 40 thousand people. Since prediabetes is very rare, look at the recall of each class, not only the accuracy. This is a teaching example and must not be used for diagnosis.

To understand each step, follow along with the project's PDF guide. The `dictionary` sheet of `diabetes_data.xlsx` explains each column.

```python
import sys
sys.path.insert(0, "../../src")  # uses the version of the library on this computer

import numpy as np
import pandas as pd

from genetic_classifier import read_excel, train_test_split, train, predict, evaluate, confusion_matrix
```

## 1) Load the data

```python
X, y = read_excel("diabetes_data.xlsx")

X_train, X_test, y_train, y_test = train_test_split(X, y, seed=0)

print("train:", X_train.shape)
print("test:", X_test.shape)
```

## 2) Train

With 3 classes there are 2 cut points. Change the values below to experiment. Use `verbose=True` to see every generation. Training can take about 20 seconds.

```python
thresholds = np.array([-1.0, 1.0])

model = train(
    X_train, y_train,
    thresholds=thresholds,
    n_chromosomes=30,
    n_mutated_genes=2,
    n_offspring=16,
    n_generations=1000,
    seed=0,
    verbose=False,
)

print("stopped at generation:", model["generation"])
print("training fitness:", round(model["fitness"], 4))
print("training accuracy:", round(model["accuracy"], 4))
```

## 3) Predict

```python
y_pred = predict(model, X_test)

print(y_pred[:10])
```

## 4) Evaluate

Here the model is tested on lessons it did not see during training.

```python
results = evaluate(y_test, y_pred, class_names=["no diabetes", "prediabetes", "diabetes"])
```
