# Mushroom classification

The model learns to tell whether a mushroom is **edible (class 1)** or **poisonous (class 0)**.

To understand each step, follow along with the project's PDF guide.

```python
import sys
sys.path.insert(0, "../../src")  # uses the version of the library on this computer

import numpy as np
import pandas as pd

from genetic_classifier import read_excel, train_test_split, train, predict, evaluate, confusion_matrix
```

## 1) Load the data

```python
X, y = read_excel("mushroom_data.xlsx")

X_train, X_test, y_train, y_test = train_test_split(X, y, seed=0)

print("train:", X_train.shape)
print("test:", X_test.shape)
```

## 2) Train

Change the values below to experiment. Use `verbose=True` to see every generation.

```python
model = train(
    X_train, y_train,
    n_chromosomes=30,
    n_mutated_genes=2,
    n_offspring=16,
    n_generations=1000,
    seed=0,
    verbose=False,
)

print("stopped at generation:", model["generation"])
print("training fitness:", round(model["fitness"], 4))
print("training accuracy:", round(model["accuracy"], 4))
```

## 3) Predict

```python
y_pred = predict(model, X_test)

print(y_pred[:10])
```

## 4) Evaluate

Here the model is tested on lessons it did not see during training.

```python
results = evaluate(y_test, y_pred, class_names=["poisonous", "edible"])
```
