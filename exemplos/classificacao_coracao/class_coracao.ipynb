# Classificação de doença cardíaca

O modelo aprende a dizer se um paciente tem **doença cardíaca (classe 1)** ou **não (classe 0)**.

Este é um exemplo didático e não serve para diagnóstico.

Para entender cada etapa, acompanhe com o guia em PDF do projeto. A aba `dicionario` de `dados_coracao.xlsx` explica cada coluna.

```python
import sys
sys.path.insert(0, "../../src")  # usa a versão da biblioteca que está neste computador

import numpy as np
import pandas as pd

from algoritmo_genetico import ler_excel, dividir_treino_teste, treinar, prever, avaliar, matriz_confusao
```

## 1) Carregar os dados

```python
X, y = ler_excel("dados_coracao.xlsx")

X_treino, X_teste, y_treino, y_teste = dividir_treino_teste(X, y, seed=0)

print("treino:", X_treino.shape)
print("teste:", X_teste.shape)
```

## 2) Treinar

Mude os valores abaixo para testar. Use `verbose=True` para ver cada geração.

```python
modelo = treinar(
    X_treino, y_treino,
    qtd_cromossomos=1000,
    qtd_genes_mutados=11,
    qtd_filhos=30,
    geracoes=1000,
    seed=0,
    verbose=True,
)

print("parou na geração:", modelo["geracao"])
print("fitness no treino:", round(modelo["fitness"], 4))
print("acurácia no treino:", round(modelo["acuracia"], 4))
```

## 3) Prever

```python
previsoes = prever(modelo, X_teste)

print(previsoes[:10])
```

## 4) Avaliar

Aqui o modelo é testado com lições que ele não viu no treino.

```python
resultado = avaliar(y_teste, previsoes, nomes_classes=["sem doença", "com doença"])
```
