# Estágios de diabetes (3 classes)

O modelo aprende a dizer se uma pessoa está **sem diabetes (classe 0)**, com **pré-diabetes (classe 1)** ou com **diabetes (classe 2)**, usando dados de saúde e hábitos de uma pesquisa.

O arquivo usa uma amostra aleatória de 40 mil pessoas. Como o pré-diabetes é muito raro, olhe o recall de cada classe, e não só a acurácia. Este é um exemplo didático e não serve para diagnóstico.

Para entender cada etapa, acompanhe com o guia em PDF, na pasta `documentacao`. A aba `dicionario` de `dados_diabetes.xlsx` explica cada coluna.

```python
import sys
sys.path.insert(0, "../../src")  # usa a versão da biblioteca que está neste computador

import numpy as np
import pandas as pd

from algoritmo_genetico import ler_excel, dividir_treino_teste, treinar, prever, avaliar, matriz_confusao
```

## 1) Carregar os dados

```python
X, y = ler_excel("dados_diabetes.xlsx")

X_treino, X_teste, y_treino, y_teste = dividir_treino_teste(X, y, seed=0)

print("treino:", X_treino.shape)
print("teste:", X_teste.shape)
```

## 2) Treinar

Com 3 classes, são 2 pontos de corte. Mude os valores abaixo para testar. Use `verbose=True` para ver cada geração. O treino pode levar cerca de 20 segundos.

```python
thresholds = np.array([-1.0, 1.0])

modelo = treinar(
    X_treino, y_treino,
    thresholds=thresholds,
    qtd_cromossomos=30,
    qtd_genes_mutados=2,
    qtd_filhos=16,
    geracoes=1000,
    seed=0,
    verbose=False,
)

print("parou na geração:", modelo["geracao"])
print("fitness no treino:", round(modelo["fitness"], 4))
print("acurácia no treino:", round(modelo["acuracia"], 4))
```

## 3) Prever

```python
previsoes = prever(modelo, X_teste)

print(previsoes[:10])
```

## 4) Avaliar

Aqui o modelo é testado com lições que ele não viu no treino.

```python
resultado = avaliar(y_teste, previsoes, nomes_classes=["sem diabetes", "pré-diabetes", "diabetes"])
```
